Section 2.1 — Creating Tensors

```python
import torch

# Create a tensor from existing data
a = torch.tensor([1, 2, 3])

# Create a tensor filled with zeros
b = torch.zeros(2, 3)

# Create a tensor filled with ones
c = torch.ones(2, 3)

# Create a tensor with random values
d = torch.rand(2, 3)

# Create a tensor from a normal distribution
e = torch.randn(2, 3)

print(f"a: {a}\n")
print(f"b: {b}\n")
print(f"c: {c}\n")
print(f"d: {d}\n")
print(f"e: {e}")
```

```python
x = torch.randn(3, 4)

print(x.shape) # Number of elements along each dimension.
print(x.ndim) # Number of dimensions (also called the tensor's rank).
print(x.dtype) # Data type of its elements, such as torch.float32.
# Note: a tensor containing integers cannot directly be configured to track gradients using requires_grad=True.
print(x.device) # Where the tensor is stored, such as cpu or mps:0.
```

```python
Z = np.array([[1, 2], [3, 4]])

A = torch.ones(20, 21)
B = torch.tensor(Z)
C = torch.rand(A.shape)
D = torch.tensor([i for i in range(4,40,2)])
```

- torch.tensor(Z) creates a copy of the NumPy data.
- torch.from_numpy(Z) creates a tensor that shares memory with the NumPy array.
- `C = torch.rand_like(A)` also works. This also inherits A's data type and device by default.
- `D = torch.arange(4, 42, 2)` also works.

```python
print(f"A: {A}\n")
print(f"B: {B}\n")
print(f"C: {C}\n")
print(f"D: {D}\n")
```

# Section 2.2: Operations in PyTorch

tensor operations

- A * B performs element-wise multiplication.
- A @ B performs matrix multiplication.

```python
a = torch.ones(5, 3)
print(f"a: {a}")
b = torch.rand(5, 3)
print(f"b: {b}")

c = torch.empty(5, 3)
print(f"c: {c}")
d = torch.empty(5, 3)
print(f"d: {d}")

torch.add(a, b, out=c) # Here, out=c means the result is written into the existing tensor c, rather than creating a new result tensor.
torch.multiply(a, b, out=d)
print(f"c: {c}")
print(f"d: {d}")
```

tensor methods

```python
x = torch.tensor([
    [1., 2., 3.],
    [4., 5., 6.]
])

print(f"sum: {x.sum()}")
print(f"sum dim=0: {x.sum(dim=0)}")
print(f"sum dim=1: {x.sum(dim=1)}")
```
