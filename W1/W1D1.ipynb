# Section 2.1 — Creating Tensors

```python
import torch

# Create a tensor from existing data
a = torch.tensor([1, 2, 3])

# Create a tensor filled with zeros
b = torch.zeros(2, 3)

# Create a tensor filled with ones
c = torch.ones(2, 3)

# Create a tensor with random values
d = torch.rand(2, 3)

# Create a tensor from a normal distribution
e = torch.randn(2, 3)

print(f"a: {a}\n")
print(f"b: {b}\n")
print(f"c: {c}\n")
print(f"d: {d}\n")
print(f"e: {e}")
```

# Section 2.2: Operations in PyTorch

tensor operations

- A * B performs element-wise multiplication.
- A @ B performs matrix multiplication.

```python
a = torch.ones(5, 3)
print(f"a: {a}")
b = torch.rand(5, 3)
print(f"b: {b}")

c = torch.empty(5, 3)
print(f"c: {c}")
d = torch.empty(5, 3)
print(f"d: {d}")

torch.add(a, b, out=c) # Here, out=c means the result is written into the existing tensor c, rather than creating a new result tensor.
torch.multiply(a, b, out=d)
print(f"c: {c}")
print(f"d: {d}")
```

tensor methods

```python
x = torch.tensor([
    [1., 2., 3.],
    [4., 5., 6.]
])

print(f"sum: {x.sum()}")
print(f"sum dim=0: {x.sum(dim=0)}")
print(f"sum dim=1: {x.sum(dim=1)}")
```

Matrix multiplication is central to neural networks, for example computing a fully connected layer:
# Y = WX + b

Dot product multiplies corresponding vector elements and sums the result:

```python
a = torch.tensor([3, 5, 7])
b = torch.tensor([2, 4, 8])

result = torch.dot(a, b)
print(result)
```

# Coding Exercise 2.2: Simple tensor operations

```python
def simple_operations(a1, a2, a3):
    answer = a1 @ a2 + a3
    return answer

a1 = torch.tensor([[2, 4], [5, 7]])
a2 = torch.tensor([[1, 1], [2, 3]])
a3 = torch.tensor([[10, 10], [12, 1]])

print(simple_operations(a1, a2, a3))
```

```python
def dot_product(b1, b2):
    product = torch.dot(b1, b2)
    return product

b1 = torch.tensor([3, 5, 7])
b2 = torch.tensor([2, 4, 8])

print(dot_product(b1, b2))
```

# 2.3 Manipulating Tensors

```python
x = torch.arange(12)
print(x.shape)

y = x.reshape(3, 4)
print(y.shape)
x, y
```

Reshaping a tensor must preserve its total number of elements.

Total number of elements is the product of all the dimensions.

Reshaping changes how elements are organised into dimensions, not the number of elements or their values. It preserves their logical order by default.

I can let PyTorch infer one dimension using -1:

```python
x = torch.arange(24)

y = x.reshape(3, -1)

print(y.shape)
```

`reshape()` vs `view()`

- `reshape()` returns the requested shape, copying data if necessary
- `view()` requires a compatible memory layout and cannot always perform the operation

`squeeze()` and `unsqueeze()`
- `unsqueeze()` adds an extra dimension of depth 1
- `squeeze()` removes a dimension of depth 1

```python
x = torch.randn(1, 32, 1, 10)
y = x.squeeze(2)
z = y.unsqueeze(0)
```

slicing

```python
x = torch.tensor([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(x[2][1])

x[0, :]    # Entire first row: [10, 20, 30]
x[:, 1]    # Entire second column: [20, 50, 80]
x[0:2, :]  # First two rows
x[:, 1:3]  # Second and third columns
```

`permute()`. Unlike `reshape()`, which reorganises elements into a new shape, `permute()` changes the order of the tensor's aces.

```python
x = torch.randn(2, 3, 4)
y = x.permute(2, 0, 1)
print(x)
print(y)

# New dimension 0 comes from old dimension 2.
# New dimension 1 comes from old dimension 0.
# New dimension 2 comes from old dimension 1.
```

Especially useful for images, since different libraries expect different dimension orders, such as (height, width, channels) vs. (channels, height, width)

Concatenating tensors: `torch.cat()` joins tensors along an existing dimension.

```python
a = torch.ones(2, 3)
b = torch.zeros(2, 3)

c = torch.cat((a, b), dim=0)
d = torch.cat((a, b), dim=1)

print(f"c: {c}")
print(f"d: {d}")
```

# Coding Exercise 2.3: Manipulating Tensors

```python
def functionA(my_tensor1, my_tensor2):
    col_sum = my_tensor1.sum(dim=0)
    t2_sum = my_tensor2.sum()
    product = col_sum * t2_sum
    output = product

    return output
```

```python
A = torch.tensor([[1, 1],
                  [1, 1]])

B = torch.tensor([[1, 2, 3],
                  [1, 2, 3]])

print(functionA(A, B))
# Expected: tensor([24, 24])
```
