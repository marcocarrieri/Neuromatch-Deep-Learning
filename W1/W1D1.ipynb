```python
import torch

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")

a = torch.randn(1000, 1000, device=device)
b = torch.randn(1000, 1000, device=device)

c = a @ b

print("Device:", c.device)
print("Result shape:", c.shape)
```
